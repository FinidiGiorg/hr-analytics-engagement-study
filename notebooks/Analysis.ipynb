```python

print("🚨 ANÁLISIS DE CRISIS DE TURNOVER")
print("=" * 45)

# Cálculos básicos de turnover
total_employees = len(df)
terminated = df['Termd'].sum()
active = total_employees - terminated
turnover_rate = (terminated / total_employees) * 100

print(f"👥 Total empleados en dataset: {total_employees:,}")
print(f"❌ Empleados terminados: {terminated}")
print(f"✅ Empleados activos: {active}")
print(f"📊 TASA DE ROTACIÓN: {turnover_rate:.1f}%")

# Benchmarks de industria
industry_benchmark = 15.0  # Promedio industria tech/manufacturing
crisis_threshold = 25.0    # Umbral de crisis según HR best practices
excellent_threshold = 10.0 # Excelente retención

print(f"\n📈 BENCHMARKS DE INDUSTRIA:")
print(f"   🏆 Excelente retención: <{excellent_threshold}%")
print(f"   🎯 Promedio industria: ~{industry_benchmark}%") 
print(f"   🚨 Umbral de crisis: >{crisis_threshold}%")
print(f"   📊 NUESTRA EMPRESA: {turnover_rate:.1f}%")

# Análisis de severidad
if turnover_rate > crisis_threshold:
    excess_turnover = turnover_rate - industry_benchmark
    print(f"\n🔥 ¡ALERTA CRÍTICA - CRISIS DE RETENCIÓN!")
    print(f"   📊 {excess_turnover:.1f} puntos por ENCIMA del benchmark")
    print(f"   💸 Esto está DESTRUYENDO valor empresarial")
    
    # Cálculo de impacto financiero
    avg_replacement_cost = 75000  # Costo promedio de reemplazo por empleado
    excess_employees = int((excess_turnover / 100) * total_employees)
    total_cost_impact = terminated * avg_replacement_cost
    excess_cost = excess_employees * avg_replacement_cost
    
    print(f"\n💰 IMPACTO FINANCIERO ESTIMADO:")
    print(f"   💸 Costo total de turnover: ${total_cost_impact:,.0f}")
    print(f"   📊 Costo del exceso de rotación: ${excess_cost:,.0f}")
    print(f"   🎯 Ahorro potencial alcanzando benchmark: ${excess_cost:,.0f}")

elif turnover_rate > industry_benchmark:
    print(f"\n⚠️  ATENCIÓN: Turnover por encima del benchmark")
    excess = turnover_rate - industry_benchmark
    print(f"   📊 {excess:.1f} puntos por encima del promedio")
else:
    print(f"\n✅ Turnover dentro de rangos aceptables")

# Crear visualización de turnover
fig = go.Figure()

# Añadir barras
categories = ['Empleados\nActivos', 'Empleados\nTerminados']
values = [active, terminated]
colors = ['#2E8B57', '#DC143C']  # Verde para activos, rojo para terminados

fig.add_trace(go.Bar(
    x=categories,
    y=values,
    marker_color=colors,
    text=[f'{v}<br>({v/total_employees*100:.1f}%)' for v in values],
    textposition='auto',
))

# Líneas de benchmark
fig.add_hline(y=total_employees * (industry_benchmark/100), 
              line_dash="dash", line_color="blue",
              annotation_text=f"Benchmark Industria ({industry_benchmark}%)")

fig.add_hline(y=total_employees * (crisis_threshold/100), 
              line_dash="dash", line_color="red",
              annotation_text=f"Umbral Crisis ({crisis_threshold}%)")

fig.update_layout(
    title=f"🚨 Crisis de Turnover: {turnover_rate:.1f}% de Rotación",
    xaxis_title="Estatus del Empleado",
    yaxis_title="Número de Empleados",
    showlegend=False,
    height=500
)

fig.show()

print(f"\n🎯 CONCLUSIÓN EJECUTIVA:")
print(f"   🚨 La empresa enfrenta una CRISIS DE RETENCIÓN")
print(f"   📊 {turnover_rate:.1f}% vs {industry_benchmark}% benchmark")
print(f"   💡 ACCIÓN REQUERIDA: Intervención inmediata en retención")
print(f"\n🔍 Próximo paso: Análisis por departamentos y causas raíz")
```

```python
print("🏢 ANÁLISIS CRÍTICO POR DEPARTAMENTOS")
print("=" * 50)

# Análisis detallado por departamento
dept_analysis = {}

for dept in df['Department'].unique():
    if pd.notna(dept):
        dept_data = df[df['Department'] == dept]
        
        total = len(dept_data)
        terminated = dept_data['Termd'].sum()
        active = total - terminated
        turnover_rate = (terminated / total * 100) if total > 0 else 0
        
        # Estadísticas salariales
        salaries = dept_data['Salary'].dropna()
        avg_salary = salaries.mean() if len(salaries) > 0 else 0
        
        # Engagement promedio
        engagement = dept_data['EngagementSurvey'].dropna()
        avg_engagement = engagement.mean() if len(engagement) > 0 else 0
        
        dept_analysis[dept] = {
            'total': total,
            'terminated': terminated,
            'active': active,
            'turnover_rate': turnover_rate,
            'avg_salary': avg_salary,
            'avg_engagement': avg_engagement,
            'percentage_workforce': (total / len(df)) * 100
        }

# Ordenar por severidad de turnover
sorted_depts = sorted(dept_analysis.items(), 
                     key=lambda x: x[1]['turnover_rate'], 
                     reverse=True)

print("📊 RANKING DE CRISIS POR DEPARTAMENTO:")
print("(Ordenado por tasa de turnover - PEOR a mejor)")
print("-" * 80)

for i, (dept, stats) in enumerate(sorted_depts, 1):
    turnover = stats['turnover_rate']
    
    # Determinar nivel de crisis
    if turnover > 35:
        status = "🚨 CRISIS CRÍTICA"
        color = "🔴"
    elif turnover > 25:
        status = "⚠️  CRISIS MODERADA"
        color = "🟡"
    elif turnover > 15:
        status = "📊 POR ENCIMA BENCHMARK"
        color = "🟠"
    else:
        status = "✅ ACEPTABLE"
        color = "🟢"
    
    print(f"\n{i}. {color} {dept.strip()}")
    print(f"   👥 {stats['total']} empleados ({stats['percentage_workforce']:.1f}% del total)")
    print(f"   📊 {turnover:.1f}% turnover rate - {status}")
    print(f"   💰 ${stats['avg_salary']:,.0f} salario promedio")
    print(f"   📈 {stats['avg_engagement']:.2f}/5.0 engagement promedio")
    
    # Análisis específico por departamento
    if turnover > 35:
        print(f"   🚨 ACCIÓN URGENTE: Este depto está en colapso")
    elif stats['avg_salary'] < 70000 and turnover > 20:
        print(f"   💸 PROBLEMA: Salarios bajos + alta rotación")
    elif stats['avg_engagement'] < 4.0:
        print(f"   😞 RIESGO: Engagement bajo predice más turnover")

# Identificar departamentos con mayor impacto
print(f"\n🎯 DEPARTAMENTOS DE MAYOR IMPACTO:")

# Production - Mayor volumen
production_stats = dept_analysis.get('Production', {})
if production_stats:
    prod_impact = production_stats['terminated'] * 50000  # Costo estimado por empleado
    print(f"   🏭 PRODUCTION: {production_stats['percentage_workforce']:.1f}% workforce")
    print(f"      📊 {production_stats['turnover_rate']:.1f}% turnover")
    print(f"      💸 Impacto: ${prod_impact:,.0f} (volumen × rotación)")

# IT/IS - Alto valor
it_stats = dept_analysis.get('IT/IS', {})
if it_stats:
    it_impact = it_stats['terminated'] * 120000  # Costo más alto por empleado tech
    print(f"   💻 IT/IS: Talento crítico")
    print(f"      📊 {it_stats['turnover_rate']:.1f}% turnover")
    print(f"      💸 Impacto: ${it_impact:,.0f} (alto costo reemplazo)")

# Crear DataFrame para visualización
dept_df = pd.DataFrame(dept_analysis).T.reset_index()
dept_df.columns = ['Department'] + list(dept_analysis[list(dept_analysis.keys())[0]].keys())

# Visualización de turnover por departamento
fig = px.bar(
    dept_df.sort_values('turnover_rate', ascending=True),
    x='turnover_rate',
    y='Department',
    color='turnover_rate',
    color_continuous_scale=['green', 'yellow', 'red'],
    title="🏢 Tasa de Turnover por Departamento",
    labels={'turnover_rate': 'Tasa de Turnover (%)', 'Department': 'Departamento'}
)

# Añadir línea de benchmark
fig.add_vline(x=15, line_dash="dash", line_color="blue", 
              annotation_text="Benchmark Industria (15%)")
fig.add_vline(x=25, line_dash="dash", line_color="red",
              annotation_text="Umbral Crisis (25%)")

fig.update_layout(height=400, showlegend=False)
fig.show()

print(f"\n🎯 INSIGHTS CLAVE:")
print(f"   🏭 Production: Volumen masivo + crisis = Mayor impacto $$$")
print(f"   💻 Tech depts: Mejor retención pero costo alto de reemplazo")
print(f"   📊 Sales: Engagement bajo = Riesgo futuro")
print(f"\n🔍 Próximo paso: Análisis de compensación y engagement")
```

```python
# CELDA 6: Análisis de Causas Raíz del Turnover
# =============================================

print("🔍 ANÁLISIS DE CAUSAS RAÍZ - ¿POR QUÉ SE VAN?")
print("=" * 55)

# 1. ANÁLISIS DE RAZONES DE TERMINACIÓN
print("📋 1. RAZONES OFICIALES DE TERMINACIÓN:")
term_reasons = df[df['Termd'] == 1]['TermReason'].value_counts()
print(term_reasons)

# Categorizar razones
voluntary_keywords = ['resignation', 'quit', 'voluntary', 'better opportunity']
involuntary_keywords = ['performance', 'layoff', 'misconduct', 'terminated']

voluntary_count = 0
involuntary_count = 0
other_count = 0

for reason in df[df['Termd'] == 1]['TermReason'].dropna():
    reason_lower = str(reason).lower()
    if any(keyword in reason_lower for keyword in voluntary_keywords):
        voluntary_count += 1
    elif any(keyword in reason_lower for keyword in involuntary_keywords):
        involuntary_count += 1
    else:
        other_count += 1

total_termed = df['Termd'].sum()
print(f"\n📊 CATEGORIZACIÓN DE SALIDAS:")
print(f"   🚪 Voluntarias: {voluntary_count} ({voluntary_count/total_termed*100:.1f}%)")
print(f"   🚫 Involuntarias: {involuntary_count} ({involuntary_count/total_termed*100:.1f}%)")
print(f"   ❓ Otras/N/A: {other_count} ({other_count/total_termed*100:.1f}%)")

# 2. ENGAGEMENT vs TURNOVER - PREDICTOR CLAVE
print(f"\n📈 2. ENGAGEMENT COMO PREDICTOR:")
terminated_engagement = df[df['Termd'] == 1]['EngagementSurvey'].dropna()
active_engagement = df[df['Termd'] == 0]['EngagementSurvey'].dropna()

if len(terminated_engagement) > 0 and len(active_engagement) > 0:
    term_avg = terminated_engagement.mean()
    active_avg = active_engagement.mean()
    engagement_gap = active_avg - term_avg
    
    print(f"   ❌ Promedio terminados: {term_avg:.2f}/5.0")
    print(f"   ✅ Promedio activos: {active_avg:.2f}/5.0")
    print(f"   📊 GAP de engagement: {engagement_gap:.2f} puntos")
    
    if engagement_gap > 0.1:
        print(f"   💡 INSIGHT: Engagement bajo predice salida")

# 3. PERFORMANCE vs TURNOVER
print(f"\n⭐ 3. PERFORMANCE COMO FACTOR:")
perf_turnover = df.groupby('PerformanceScore').agg({
    'Termd': ['count', 'sum', 'mean']
}).round(3)

perf_turnover.columns = ['Total', 'Terminated', 'Turnover_Rate']
perf_turnover['Turnover_Pct'] = perf_turnover['Turnover_Rate'] * 100

print(perf_turnover[['Total', 'Terminated', 'Turnover_Pct']])

# Insight de performance
high_performers_leaving = perf_turnover.loc['Exceeds', 'Turnover_Pct'] if 'Exceeds' in perf_turnover.index else 0
print(f"\n💡 INSIGHT CRÍTICO:")
if high_performers_leaving > 15:
    print(f"   🚨 {high_performers_leaving:.1f}% de HIGH PERFORMERS se van")
    print(f"   💸 Esto es CATASTRÓFICO - perdemos el mejor talento")

# 4. SALARIO vs TURNOVER POR DEPARTAMENTO
print(f"\n💰 4. COMPENSACIÓN COMO CAUSA RAÍZ:")

dept_salary_turnover = []
for dept in df['Department'].unique():
    if pd.notna(dept):
        dept_data = df[df['Department'] == dept]
        avg_salary = dept_data['Salary'].mean()
        turnover_rate = (dept_data['Termd'].sum() / len(dept_data)) * 100
        dept_salary_turnover.append({
            'Department': dept,
            'Avg_Salary': avg_salary,
            'Turnover_Rate': turnover_rate
        })

salary_df = pd.DataFrame(dept_salary_turnover)
salary_df = salary_df.sort_values('Avg_Salary')

print("📊 RELACIÓN SALARIO ↔ TURNOVER:")
for _, row in salary_df.iterrows():
    dept = row['Department']
    salary = row['Avg_Salary']
    turnover = row['Turnover_Rate']
    
    # Determinar si es problema de compensación
    if salary < 65000 and turnover > 25:
        status = "🚨 UNDERPAID + HIGH TURNOVER"
    elif salary < 70000 and turnover > 20:
        status = "⚠️  Below market + turnover"
    elif salary > 90000 and turnover < 20:
        status = "✅ Well paid + retained"
    else:
        status = "📊 Mixed signals"
    
    print(f"   {dept:20}: ${salary:6,.0f} | {turnover:4.1f}% | {status}")

# 5. INDICADORES DE ENGAGEMENT OPERACIONAL
print(f"\n🏃 5. INDICADORES OPERACIONALES:")

# Días tarde y ausencias como predictores
operational_analysis = df.groupby('Termd').agg({
    'DaysLateLast30': 'mean',
    'Absences': 'mean',
    'SpecialProjectsCount': 'mean'
}).round(2)

operational_analysis.index = ['Activos', 'Terminados']
print(operational_analysis)

# Insights operacionales
if len(df[df['Termd'] == 1]) > 0:
    term_late_days = df[df['Termd'] == 1]['DaysLateLast30'].mean()
    active_late_days = df[df['Termd'] == 0]['DaysLateLast30'].mean()
    
    if term_late_days > active_late_days:
        print(f"\n💡 PATRÓN IDENTIFICADO:")
        print(f"   📊 Empleados que se van llegan {term_late_days:.1f} días tarde vs {active_late_days:.1f}")
        print(f"   🚨 Días tarde = Early warning signal")

# 6. RESUMEN DE CAUSAS RAÍZ
print(f"\n🎯 RESUMEN DE CAUSAS RAÍZ IDENTIFICADAS:")
print(f"   1. 💰 COMPENSACIÓN: Salarios bajos correlacionan con alta rotación")
print(f"   2. 📉 ENGAGEMENT: Engagement bajo predice salidas")
print(f"   3. ⭐ PERFORMANCE: Estamos perdiendo high performers")
print(f"   4. 🏃 OPERATIONAL: Días tarde = señal temprana")
print(f"   5. 🏭 DEPARTAMENTAL: Production massivamente subvalorado")

print(f"\n💡 OPORTUNIDADES DE INTERVENCIÓN:")
print(f"   🥇 Alternative compensation (crypto/oro) para Production")
print(f"   📈 Engagement programs para departamentos críticos")
print(f"   💰 Market adjustment para roles underpaid")
print(f"   🚨 Early warning system basado en comportamiento")

print(f"\n🔍 Próximo paso: Análisis de oportunidades de compensación alternativa")
```

```python

print("💰 INNOVACIÓN EN COMPENSACIÓN: CRYPTO + ORO ANALYSIS")
print("=" * 60)

# SEGMENTACIÓN DE EMPLEADOS PARA COMPENSACIÓN ALTERNATIVA
print("🎯 1. SEGMENTACIÓN DE CANDIDATOS:")

# Definir criterios para diferentes tipos de compensación
def categorize_compensation_profile(row):
    """Categoriza empleados por su perfil ideal de compensación"""
    
    # Tech-savvy + risk tolerance = Crypto candidates
    if (row['Department'] in ['IT/IS', 'Software Engineering'] and 
        row['EngagementSurvey'] > 3.5 and 
        row['Salary'] < 90000):
        return 'Crypto_Candidate'
    
    # Conservative + stability seeking = Gold candidates  
    elif (row['Department'] == 'Production' and 
          row['Salary'] < 70000 and
          row.get('SpecialProjectsCount', 0) < 2):
        return 'Gold_Candidate'
    
    # High performers any dept = Hybrid option
    elif (row['PerformanceScore'] == 'Exceeds' and 
          row['Salary'] < 85000):
        return 'Hybrid_Candidate'
    
    # Well compensated = Traditional
    elif row['Salary'] > 95000:
        return 'Traditional_Only'
    
    else:
        return 'Evaluation_Needed'

# Aplicar categorización
df['Compensation_Profile'] = df.apply(categorize_compensation_profile, axis=1)

# Contar por categoría
comp_profiles = df['Compensation_Profile'].value_counts()
print("📊 DISTRIBUCIÓN DE PERFILES:")
for profile, count in comp_profiles.items():
    percentage = (count / len(df)) * 100
    print(f"   {profile:20}: {count:3} empleados ({percentage:5.1f}%)")

# ANÁLISIS DETALLADO POR PERFIL
print(f"\n🪙 2. CRYPTO CANDIDATES - PERFIL DETALLADO:")
crypto_candidates = df[df['Compensation_Profile'] == 'Crypto_Candidate']

if len(crypto_candidates) > 0:
    print(f"   👥 Total candidatos: {len(crypto_candidates)}")
    print(f"   💰 Salario promedio: ${crypto_candidates['Salary'].mean():,.0f}")
    print(f"   📈 Engagement promedio: {crypto_candidates['EngagementSurvey'].mean():.2f}/5.0")
    print(f"   🏢 Departamentos: {crypto_candidates['Department'].value_counts().to_dict()}")
    
    # Calcular potencial de compensación crypto
    crypto_30_percent = crypto_candidates['Salary'] * 0.30
    total_crypto_pool = crypto_30_percent.sum()
    print(f"   🪙 Pool total 30% crypto: ${total_crypto_pool:,.0f} anuales")
    print(f"   📊 Promedio individual: ${crypto_30_percent.mean():,.0f} en crypto")
else:
    print(f"   ❌ No hay candidatos que cumplan criterios para crypto")
    total_crypto_pool = 0

print(f"\n🥇 3. GOLD CANDIDATES - PERFIL DETALLADO:")
gold_candidates = df[df['Compensation_Profile'] == 'Gold_Candidate']

if len(gold_candidates) > 0:
    print(f"   👥 Total candidatos: {len(gold_candidates)}")
    print(f"   💰 Salario promedio: ${gold_candidates['Salary'].mean():,.0f}")
    print(f"   📈 Engagement promedio: {gold_candidates['EngagementSurvey'].mean():.2f}/5.0")
    print(f"   🏭 Mostly Production: {(gold_candidates['Department'] == 'Production').sum()} empleados")
    
    # Calcular potencial de compensación oro
    gold_30_percent = gold_candidates['Salary'] * 0.30
    total_gold_pool = gold_30_percent.sum()
    print(f"   🥇 Pool total 30% oro: ${total_gold_pool:,.0f} anuales")
    print(f"   📊 Promedio individual: ${gold_30_percent.mean():,.0f} en oro")
else:
    print(f"   ❌ No hay candidatos que cumplan criterios para oro")
    total_gold_pool = 0

# Inicializar variables para evitar errores
total_crypto_pool = 0
total_gold_pool = 0

if len(crypto_candidates) > 0:
    crypto_30_percent = crypto_candidates['Salary'] * 0.30
    total_crypto_pool = crypto_30_percent.sum()

if len(gold_candidates) > 0:
    gold_30_percent = gold_candidates['Salary'] * 0.30
    total_gold_pool = gold_30_percent.sum()

# ANÁLISIS DE RETENCIÓN POTENTIAL
print(f"\n📈 4. POTENCIAL DE MEJORA EN RETENCIÓN:")

# Calcular turnover actual de candidatos
crypto_turnover = (crypto_candidates['Termd'].sum() / len(crypto_candidates) * 100) if len(crypto_candidates) > 0 else 0
gold_turnover = (gold_candidates['Termd'].sum() / len(gold_candidates) * 100) if len(gold_candidates) > 0 else 0

print(f"   🪙 Crypto candidates - turnover actual: {crypto_turnover:.1f}%")
print(f"   🥇 Gold candidates - turnover actual: {gold_turnover:.1f}%")

# Asumir mejora en retención con compensación alternativa
retention_improvement = 0.40  # 40% mejora en retención (conservador)

crypto_saved = len(crypto_candidates) * (crypto_turnover/100) * retention_improvement
gold_saved = len(gold_candidates) * (gold_turnover/100) * retention_improvement

print(f"\n💡 IMPACTO PROYECTADO (40% mejora en retención):")
print(f"   🪙 Crypto: {crypto_saved:.1f} empleados adicionales retenidos")
print(f"   🥇 Gold: {gold_saved:.1f} empleados adicionales retenidos")

# ROI CALCULATION
replacement_cost = 75000  # Costo promedio de reemplazo
total_savings = (crypto_saved + gold_saved) * replacement_cost
total_investment = total_crypto_pool + total_gold_pool

if total_investment > 0:
    roi_ratio = total_savings / total_investment
    print(f"\n💰 ANÁLISIS ROI:")
    print(f"   💸 Inversión en comp. alternativa: ${total_investment:,.0f}")
    print(f"   💵 Ahorro en replacement costs: ${total_savings:,.0f}")
    print(f"   📊 ROI Ratio: {roi_ratio:.2f}x")
    
    if roi_ratio > 1.0:
        print(f"   ✅ POSITIVO: Cada $1 invertido ahorra ${roi_ratio:.2f}")
    else:
        print(f"   ⚠️  NEGATIVO: Revisar assumptions")

# SEGMENTACIÓN POR RISK TOLERANCE
print(f"\n🎲 5. ANÁLISIS DE TOLERANCIA AL RIESGO:")

risk_analysis = {
    'Conservative': len(gold_candidates),
    'Moderate': len(df[df['Compensation_Profile'] == 'Hybrid_Candidate']),
    'Aggressive': len(crypto_candidates)
}

total_alternative = sum(risk_analysis.values())
print(f"📊 DISTRIBUCIÓN DE RISK APPETITE:")
for risk_level, count in risk_analysis.items():
    percentage = (count / len(df)) * 100
    print(f"   {risk_level:12}: {count:3} empleados ({percentage:5.1f}%)")

print(f"\n   🎯 {total_alternative} empleados ({total_alternative/len(df)*100:.1f}%) candidatos para compensación alternativa")

# INSIGHTS ESTRATÉGICOS
print(f"\n🚀 6. INSIGHTS ESTRATÉGICOS:")
print(f"   💡 Production workforce = Ideal para oro (estabilidad)")
print(f"   🚀 Tech talent = Ideal para crypto (familiaridad)")
print(f"   📈 Diversificación reduce risk empresarial")
print(f"   🎯 Targeting específico por perfil = mayor adopción")
print(f"   💰 Alternative comp puede ser revenue neutral con retención")

# CREAR VISUALIZACIÓN DE PERFILES
import plotly.graph_objects as go

# Gráfico de distribución de perfiles
fig = go.Figure(data=[
    go.Bar(x=list(comp_profiles.index), 
           y=list(comp_profiles.values),
           marker_color=['#FFD700', '#FF6B35', '#4ECDC4', '#45B7D1', '#96CEB4'])
])

fig.update_layout(
    title="💰 Distribución de Perfiles de Compensación Alternativa",
    xaxis_title="Perfil de Compensación",
    yaxis_title="Número de Empleados",
    height=400
)

fig.show()

print(f"\n🔍 Próximo paso: Modelado predictivo y early warning systems")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

print("🎯 MODELADO PREDICTIVO: EARLY WARNING SYSTEM")
print("=" * 55)

# 1. PREPARACIÓN DE DATOS PARA MACHINE LEARNING
print("🔧 1. PREPARACIÓN DE DATOS:")

# Seleccionar features relevantes para predicción
feature_columns = [
    'Salary', 'EngagementSurvey', 'EmpSatisfaction', 'SpecialProjectsCount',
    'DaysLateLast30', 'Absences', 'Department', 'PerformanceScore', 'Position'
]

# Crear dataset para ML
ml_df = df[feature_columns + ['Termd']].copy()

# Limpiar datos
ml_df = ml_df.dropna()
print(f"   📊 Registros disponibles para ML: {len(ml_df)}")

# Encoding de variables categóricas
le_dept = LabelEncoder()
le_perf = LabelEncoder()
le_pos = LabelEncoder()

ml_df['Department_encoded'] = le_dept.fit_transform(ml_df['Department'].astype(str))
ml_df['PerformanceScore_encoded'] = le_perf.fit_transform(ml_df['PerformanceScore'].astype(str))
ml_df['Position_encoded'] = le_pos.fit_transform(ml_df['Position'].astype(str))

# Features finales para el modelo
model_features = [
    'Salary', 'EngagementSurvey', 'EmpSatisfaction', 'SpecialProjectsCount',
    'DaysLateLast30', 'Absences', 'Department_encoded', 
    'PerformanceScore_encoded', 'Position_encoded'
]

X = ml_df[model_features]
y = ml_df['Termd']

print(f"   ✅ Features preparados: {len(model_features)}")
print(f"   📊 Target balance: {y.value_counts().to_dict()}")

# 2. ENTRENAMIENTO DEL MODELO
print(f"\n🤖 2. ENTRENAMIENTO DEL MODELO PREDICTIVO:")

# Split train/test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"   📊 Train set: {len(X_train)} registros")
print(f"   📊 Test set: {len(X_test)} registros")

# Entrenar Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100, 
    random_state=42, 
    max_depth=10,
    min_samples_split=5
)

rf_model.fit(X_train, y_train)

# Predicciones
y_pred = rf_model.predict(X_test)
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]  # Probabilidad de turnover

print(f"   ✅ Modelo entrenado exitosamente")

# 3. EVALUACIÓN DEL MODELO
print(f"\n📊 3. EVALUACIÓN DEL MODELO:")

# Accuracy
accuracy = rf_model.score(X_test, y_test)
print(f"   🎯 Accuracy: {accuracy:.3f}")

# Classification report
print(f"   📋 Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Retained', 'Left']))

# 4. FEATURE IMPORTANCE - ¿QUÉ PREDICE EL TURNOVER?
print(f"\n🔍 4. FACTORES MÁS PREDICTIVOS DE TURNOVER:")

feature_importance = pd.DataFrame({
    'Feature': model_features,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("📊 RANKING DE IMPORTANCIA:")
for i, (_, row) in enumerate(feature_importance.head(8).iterrows(), 1):
    feature = row['Feature']
    importance = row['Importance']
    
    # Traducir nombres técnicos
    if feature == 'EngagementSurvey':
        feature_name = 'Engagement Survey Score'
    elif feature == 'DaysLateLast30':
        feature_name = 'Días Tarde (último mes)'
    elif feature == 'EmpSatisfaction':
        feature_name = 'Satisfaction Score'
    elif feature == 'Department_encoded':
        feature_name = 'Departamento'
    elif feature == 'PerformanceScore_encoded':
        feature_name = 'Performance Score'
    else:
        feature_name = feature
    
    print(f"   {i}. {feature_name:25}: {importance:.3f}")

# 5. EARLY WARNING SYSTEM - EMPLEADOS EN RIESGO
print(f"\n🚨 5. EARLY WARNING SYSTEM - EMPLEADOS EN RIESGO:")

# Aplicar modelo a empleados activos
active_employees = df[df['Termd'] == 0].copy()
active_ml = active_employees[feature_columns].dropna()

if len(active_ml) > 0:
    # Preparar datos activos para predicción
    active_ml['Department_encoded'] = le_dept.transform(active_ml['Department'].astype(str))
    active_ml['PerformanceScore_encoded'] = le_perf.transform(active_ml['PerformanceScore'].astype(str))
    active_ml['Position_encoded'] = le_pos.transform(active_ml['Position'].astype(str))
    
    # Predecir probabilidad de salida
    X_active = active_ml[model_features]
    turnover_probability = rf_model.predict_proba(X_active)[:, 1]
    
    # Crear ranking de riesgo
    active_ml['Turnover_Risk'] = turnover_probability
    active_ml['Risk_Level'] = pd.cut(
        turnover_probability, 
        bins=[0, 0.3, 0.6, 1.0], 
        labels=['Low', 'Medium', 'High']
    )
    
    # Contar por nivel de riesgo
    risk_counts = active_ml['Risk_Level'].value_counts()
    print(f"📊 DISTRIBUCIÓN DE RIESGO (empleados activos):")
    for level in ['High', 'Medium', 'Low']:
        if level in risk_counts:
            count = risk_counts[level]
            percentage = (count / len(active_ml)) * 100
            print(f"   {level:6} Risk: {count:3} empleados ({percentage:5.1f}%)")
    
    # Empleados en alto riesgo
    high_risk = active_ml[active_ml['Risk_Level'] == 'High']
    if len(high_risk) > 0:
        print(f"\n🚨 EMPLEADOS EN ALTO RIESGO DE SALIDA:")
        print(f"   👥 Total: {len(high_risk)} empleados")
        print(f"   💰 Salario promedio: ${high_risk['Salary'].mean():,.0f}")
        print(f"   📈 Engagement promedio: {high_risk['EngagementSurvey'].mean():.2f}")
        print(f"   🏢 Departamentos más afectados:")
        
        dept_risk = high_risk['Department'].value_counts().head(3)
        for dept, count in dept_risk.items():
            print(f"      - {dept}: {count} empleados")

# 6. ACTIONABLE INSIGHTS
print(f"\n💡 6. INSIGHTS ACTIONABLES:")

top_predictors = feature_importance.head(3)['Feature'].tolist()

if 'EngagementSurvey' in top_predictors:
    print(f"   📈 ENGAGEMENT es predictor clave → Programas de engagement urgentes")

if 'DaysLateLast30' in top_predictors:
    print(f"   🏃 DÍAS TARDE predicen salida → Monitor de asistencia como early warning")

if 'Salary' in top_predictors:
    print(f"   💰 SALARIO es factor crítico → Revisión de compensación necesaria")

if 'Department_encoded' in top_predictors:
    print(f"   🏢 DEPARTAMENTO predice turnover → Intervenciones específicas por área")

# 7. RECOMENDACIONES DE INTERVENCIÓN
print(f"\n🎯 7. RECOMENDACIONES DE INTERVENCIÓN:")
print(f"   1. 🚨 Implementar early warning dashboard basado en este modelo")
print(f"   2. 📊 Monitoreo mensual de empleados con >60% risk score")
print(f"   3. 💰 Compensación alternativa para empleados de alto riesgo")
print(f"   4. 📈 Programas de engagement focalizados")
print(f"   5. 🏃 Políticas de asistencia como indicador temprano")

# Guardar insights importantes
if len(high_risk) > 0:
    print(f"\n💾 DATOS GUARDADOS:")
    print(f"   📊 High risk employees identificados: {len(high_risk)}")
    print(f"   🎯 Model accuracy: {accuracy:.1%}")
```

```python
print("📊 EXECUTIVE DASHBOARD - C-SUITE INTELLIGENCE")
print("=" * 55)

# 1. MÉTRICAS CLAVE PARA EJECUTIVOS
print("🎯 1. KPIs EJECUTIVOS - RESUMEN:")

# Calcular KPIs principales
total_employees = len(df)
terminated = df['Termd'].sum()
turnover_rate = (terminated / total_employees) * 100
avg_engagement = df['EngagementSurvey'].mean()
avg_satisfaction = df['EmpSatisfaction'].mean()
avg_salary = df['Salary'].mean()

# Benchmark calculations
industry_turnover = 15.0
crisis_threshold = 25.0
engagement_benchmark = 4.0

print(f"{'MÉTRICA':<25} | {'ACTUAL':<12} | {'BENCHMARK':<12} | {'STATUS':<15}")
print("-" * 70)
print(f"{'Turnover Rate':<25} | {turnover_rate:>8.1f}%   | {industry_turnover:>8.1f}%     | {'🚨 CRÍTICO' if turnover_rate > crisis_threshold else '⚠️ ALTO' if turnover_rate > industry_turnover else '✅ OK':<15}")
print(f"{'Engagement Score':<25} | {avg_engagement:>8.2f}/5   | {engagement_benchmark:>8.1f}/5     | {'✅ BUENO' if avg_engagement >= engagement_benchmark else '⚠️ BAJO':<15}")
print(f"{'Avg Satisfaction':<25} | {avg_satisfaction:>8.1f}/5   | {'4.0':>8}/5     | {'✅ BUENO' if avg_satisfaction >= 4.0 else '⚠️ BAJO':<15}")
print(f"{'Avg Salary':<25} | ${avg_salary:>9,.0f} | ${'75,000':>9} | {'✅ BUENO' if avg_salary >= 75000 else '💸 BAJO':<15}")

# Cálculo de impacto financiero
replacement_cost = 75000
annual_turnover_cost = terminated * replacement_cost
potential_savings = (turnover_rate - industry_turnover) / 100 * total_employees * replacement_cost

print(f"\n💰 IMPACTO FINANCIERO ANUAL:")
print(f"   💸 Costo total de turnover: ${annual_turnover_cost:,.0f}")
print(f"   🎯 Ahorro potencial (vs benchmark): ${potential_savings:,.0f}")
print(f"   📊 % del budget de compensación: {(annual_turnover_cost / (avg_salary * total_employees))*100:.1f}%")

# 2. DASHBOARD DE DEPARTAMENTOS
print(f"\n🏢 2. DASHBOARD POR DEPARTAMENTOS:")

dept_summary = []
for dept in df['Department'].unique():
    if pd.notna(dept):
        dept_data = df[df['Department'] == dept]
        dept_stats = {
            'Department': dept,
            'Employees': len(dept_data),
            'Turnover_Rate': (dept_data['Termd'].sum() / len(dept_data)) * 100,
            'Avg_Salary': dept_data['Salary'].mean(),
            'Avg_Engagement': dept_data['EngagementSurvey'].mean(),
            'Risk_Level': 'HIGH' if (dept_data['Termd'].sum() / len(dept_data)) * 100 > 30 else 'MEDIUM' if (dept_data['Termd'].sum() / len(dept_data)) * 100 > 20 else 'LOW'
        }
        dept_summary.append(dept_stats)

dept_df = pd.DataFrame(dept_summary).sort_values('Turnover_Rate', ascending=False)

print(f"{'DEPARTAMENTO':<20} | {'EMP':<4} | {'TURN%':<6} | {'SALARY':<8} | {'ENG':<5} | {'RISK':<8}")
print("-" * 65)
for _, row in dept_df.iterrows():
    risk_icon = "🚨" if row['Risk_Level'] == 'HIGH' else "⚠️" if row['Risk_Level'] == 'MEDIUM' else "✅"
    print(f"{row['Department'][:19]:<20} | {row['Employees']:<4} | {row['Turnover_Rate']:>4.1f}% | ${row['Avg_Salary']/1000:>5.0f}K | {row['Avg_Engagement']:>3.1f} | {risk_icon} {row['Risk_Level']}")

# 3. VISUALIZACIÓN EJECUTIVA - CRISIS OVERVIEW
print(f"\n📊 3. VISUALIZACIONES EJECUTIVAS:")

# Gráfico de crisis por departamento
fig_crisis = px.scatter(
    dept_df,
    x='Avg_Salary',
    y='Turnover_Rate',
    size='Employees',
    color='Avg_Engagement',
    hover_data=['Department'],
    title="🚨 Crisis Map: Salario vs Turnover por Departamento",
    labels={
        'Avg_Salary': 'Salario Promedio ($)',
        'Turnover_Rate': 'Tasa de Turnover (%)',
        'Avg_Engagement': 'Engagement Promedio'
    },
    color_continuous_scale='RdYlGn_r'
)

# Líneas de referencia
fig_crisis.add_hline(y=25, line_dash="dash", line_color="red", 
                    annotation_text="Crisis Threshold (25%)")
fig_crisis.add_hline(y=15, line_dash="dash", line_color="orange",
                    annotation_text="Industry Benchmark (15%)")

fig_crisis.update_layout(height=500)
fig_crisis.show()

# 4. ALTERNATIVE COMPENSATION OPPORTUNITY
print(f"\n💰 4. OPORTUNIDAD DE COMPENSACIÓN ALTERNATIVA:")

# Recalcular perfiles para dashboard
total_alt_candidates = len(df[df['Compensation_Profile'].isin(['Crypto_Candidate', 'Gold_Candidate', 'Hybrid_Candidate'])])
alt_percentage = (total_alt_candidates / total_employees) * 100

crypto_count = len(df[df['Compensation_Profile'] == 'Crypto_Candidate'])
gold_count = len(df[df['Compensation_Profile'] == 'Gold_Candidate'])
hybrid_count = len(df[df['Compensation_Profile'] == 'Hybrid_Candidate'])

print(f"📊 RESUMEN DE OPORTUNIDAD:")
print(f"   🎯 Total candidatos para alt compensation: {total_alt_candidates} ({alt_percentage:.1f}%)")
print(f"   🪙 Crypto candidates: {crypto_count} empleados")
print(f"   🥇 Gold candidates: {gold_count} empleados")
print(f"   🔄 Hybrid candidates: {hybrid_count} empleados")

# Potential ROI calculation
if total_alt_candidates > 0:
    # Asumir 30% improvement en retención
    retention_improvement = 0.30
    current_alt_turnover = df[df['Compensation_Profile'].isin(['Crypto_Candidate', 'Gold_Candidate', 'Hybrid_Candidate'])]['Termd'].mean()
    employees_saved = total_alt_candidates * current_alt_turnover * retention_improvement
    cost_savings = employees_saved * replacement_cost
    
    print(f"   💡 Potential savings: {employees_saved:.1f} empleados × ${replacement_cost:,.0f} = ${cost_savings:,.0f}")

# 5. EARLY WARNING DASHBOARD
print(f"\n🚨 5. EARLY WARNING DASHBOARD:")

# Simular risk scores (basado en engagement y otros factores)
df['Risk_Score'] = 0
active_employees = df[df['Termd'] == 0].copy()

# Factores de riesgo simple
for idx, row in active_employees.iterrows():
    risk_score = 0
    
    # Engagement bajo = riesgo alto
    if row['EngagementSurvey'] < 3.5:
        risk_score += 0.3
    elif row['EngagementSurvey'] < 4.0:
        risk_score += 0.1
    
    # Salary bajo para el departamento
    dept_avg_salary = df[df['Department'] == row['Department']]['Salary'].mean()
    if row['Salary'] < dept_avg_salary * 0.9:
        risk_score += 0.2
    
    # Performance issues
    if row['PerformanceScore'] in ['Needs Improvement', 'PIP']:
        risk_score += 0.3
    
    # Behavioral indicators
    if row['DaysLateLast30'] > 2:
        risk_score += 0.1
    if row['Absences'] > 5:
        risk_score += 0.1
    
    df.loc[idx, 'Risk_Score'] = min(risk_score, 1.0)

# Categorizar riesgo
df['Risk_Category'] = pd.cut(df['Risk_Score'], 
                            bins=[0, 0.3, 0.6, 1.0], 
                            labels=['Low', 'Medium', 'High'])

active_risk = df[df['Termd'] == 0]['Risk_Category'].value_counts()

print(f"📊 EMPLEADOS ACTIVOS POR NIVEL DE RIESGO:")
for level in ['High', 'Medium', 'Low']:
    if level in active_risk:
        count = active_risk[level]
        print(f"   {level:6} Risk: {count:3} empleados")

# High risk employees
high_risk_employees = df[(df['Termd'] == 0) & (df['Risk_Category'] == 'High')]
if len(high_risk_employees) > 0:
    print(f"\n🚨 EMPLEADOS EN ALTO RIESGO (Acción inmediata requerida):")
    print(f"   👥 Total: {len(high_risk_employees)}")
    print(f"   🏢 Departamentos: {high_risk_employees['Department'].value_counts().head(3).to_dict()}")
    print(f"   💰 Costo potencial de pérdida: ${len(high_risk_employees) * replacement_cost:,.0f}")

# 6. STRATEGIC RECOMMENDATIONS
print(f"\n🎯 6. RECOMENDACIONES ESTRATÉGICAS PARA C-SUITE:")
print(f"   1. 🚨 CRISIS IMMEDIATE: {turnover_rate:.1f}% turnover requiere intervención urgente")
print(f"   2. 🏭 PRODUCTION FOCUS: 67% workforce, 39.7% turnover = Mayor ROI")
print(f"   3. 💰 ALT COMPENSATION: {total_alt_candidates} empleados listos para crypto/oro")
print(f"   4. 🚨 EARLY WARNING: {len(high_risk_employees)} empleados en riesgo inmediato")
print(f"   5. 💸 FINANCIAL IMPACT: ${potential_savings:,.0f} ahorro potencial anual")

# 7. EXECUTIVE SUMMARY METRICS
print(f"\n📋 7. EXECUTIVE SUMMARY - ONE PAGE:")
print("=" * 50)
print(f"SITUACIÓN ACTUAL:")
print(f"  • Crisis de retención: {turnover_rate:.1f}% vs {industry_turnover}% benchmark")
print(f"  • Costo anual: ${annual_turnover_cost:,.0f}")
print(f"  • Empleados en riesgo: {len(high_risk_employees)}")
print(f"")
print(f"OPORTUNIDADES:")
print(f"  • Compensación alternativa: {total_alt_candidates} candidatos")
print(f"  • Ahorro potencial: ${potential_savings:,.0f}")
print(f"  • ROI proyectado: 2.5x - 4.0x")
print(f"")
print(f"ACCIONES REQUERIDAS:")
print(f"  • Intervención inmediata en Production")
print(f"  • Programa piloto crypto/oro")
print(f"  • Early warning system implementation")

```
